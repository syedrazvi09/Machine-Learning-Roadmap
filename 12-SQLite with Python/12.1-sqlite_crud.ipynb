```python
import sqlite3
```

```python
# connect to sqlite db
connection = sqlite3.connect('sample.db')
connection
```

```python
cursor = connection.cursor()
```

```python
# create a table
cursor.execute('''
create table if not exists students(
               id Integer Primary Key,
               name Text Not Null,
               age Integer,
               branch text)
''')

# commit the changes
connection.commit()
```

```python
cursor.execute('''
select * from students
''')
```

```python
# insert data in table
cursor.execute(
    '''
INSERT INTO students(name, age, branch)
VALUES ('Bruce', 44, 'Wayne Ind'),
    ('Harvey', 35, 'Law')
'''
)
connection.commit()
```

```python
# query the data
cursor.execute('SELECT * from students')
rows = cursor.fetchall()

# print

for row in rows:
    print(row)
```

```python
# update 
cursor.execute('''
UPDATE students
SET age = 25
WHERE name = "Bruce"
''')

connection.commit()
```

```python
# delete
cursor.execute('''
DELETE from students
               WHERE name = "Harvey"
''')

connection.commit()
```
