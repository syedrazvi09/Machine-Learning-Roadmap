```python
import threading
import time
```

```python
def print_nums():
    for i in range(4):
        time.sleep(1)
        print(f"Number: {i}")

def print_alpha():
    for alpha in 'abcd':
        time.sleep(1)
        print(f"Alphabet: {alpha}")

# create threads
t1 = threading.Thread(target = print_nums)
t2 = threading.Thread(target = print_alpha)

t = time.time()
# start thread
t1.start()
t2.start()

# wait for threads to complete
t1.join()
t2.join()

finished_time = float(time.time() - t)
print(finished_time)
```
