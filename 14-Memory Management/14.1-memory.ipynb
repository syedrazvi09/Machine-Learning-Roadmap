```python
class MyObject:
    def __init__(self, name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")
    
# create circular references
obj1 = MyObject('obj1')
obj2 = MyObject('obj2')
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

# manually trigger garbage collection
gc.collect()
```

```python
# generators for memory efficiency

def gen_nums(n):
    for i in range(n):
        yield i

for num in gen_nums(100):
    print(num)
    if num > 10:
        break
```
