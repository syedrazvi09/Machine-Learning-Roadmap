```python
class Car:
    def __init__(self, windows, doors, engine):
        self.windows = windows
        self.doors = doors
        self.engine = engine

    def drive(self):
        print('The guy will drive: ', self.engine)
        
```

```python
car1 = Car(4, 5, 'Diesel')
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self, windows, doors, engine, is_selfdriving):
        super().__init__(windows, doors, engine)
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f"Tesla can self drive: {self.is_selfdriving}")
```

```python
tesla1 = Tesla(4,6, 'Electric', True)
tesla1.selfdriving()
```

```python
tesla1.drive()
```

```python
# multiple inheritance
# when a class inherits from more than one base class

# base class 1
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        print('Subclasses implement this method')


# base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner
    
# Derived class
class Cat(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f'{self.name} says meow'
    

# create object

cat = Cat('Luna', 'Bruce')
print(cat.speak())
print(f'Owner: {cat.owner}')
```
