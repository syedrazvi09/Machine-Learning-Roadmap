```python
import pandas as pd
```

```python
# Series

data = [1,2,3,4,5]
series = pd.Series(data)
print('Series \n', series)
print(type(series))
```

```python
# series from dictionary

data = {'a':10,
        'b':20,
        'c':30}

series_dict = pd.Series(data)
print('Series From Dictionary \n', series_dict)
```

```python
data = [1,3,5,7,9]
index = ['a','b','c','d','e']

pd.Series(data, index = index)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data, index=index)
```

```python
# DataFrame

# creating a dataframe from dictionary

data = {
    'Name' : ['Bruce', 'Clark', 'Harvey'],
    'Age' : [34,67,54],
    'City' : ['Gotham', 'Metropolis', 'NYC']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create a Data Frame from a List of Dictionaries

data = [
    {'Name' :'Bruce', 'Age' : 35, 'City' : 'Gotham'},
    {'Name' :'Thanos', 'Age' : 124, 'City' : 'Titan'},
    {'Name' :'Thor', 'Age' : 500, 'City' : 'Asgard'},

]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
