```python
import seaborn as sns
```

```python
# basic plots 
planet = sns.load_dataset('planets')
```

```python
planet
```

```python
# scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x = 'distance', y = 'year', data =planet)
plt.title('Plot of planets')
plt.show()
```

```python
# line plot

sns.lineplot(x = 'distance', y = 'year', data = planet)
plt.title('Line plot')
plt.show()
```

```python
# categorical plot
# bar plot

sns.barplot(x = 'method', y = 'number', data = planet)
plt.title('Bar plot')
plt.show()
```

```python
# box plots
sns.boxplot(x = 'method', y = 'orbital_period', data = planet)
plt.show()
```

```python
# violin plot
sns.violinplot(x = 'method', y = 'distance', data = planet)
```

```python
# histograms
sns.histplot(x = 'year', y = 'distance', data = planet)
```

```python
# KDE plot
sns.kdeplot(planet['year'], fill = True)
```

```python
# pairplot
sns.pairplot(planet)
```

```python
# heatmap
corr = planet[['distance','mass','orbital_period']].corr()
corr
```

```python
sns.heatmap(corr, annot=True)
```
