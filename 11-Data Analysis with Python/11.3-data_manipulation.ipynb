```python
df['New Reading'] = df['bmi'].apply(lambda x:x*2)
df.head(5)
```

```python
# data aggration and grouping

grouped_mean = df.groupby('bmi')['age'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['age', 'region'])['bmi'].sum()
print(grouped_sum)
```

```python
grouped_agg = df.groupby('region')['bmi'].agg(['mean', 'sum', 'count'])
grouped_agg
```

```python
# merging and joining data frames
df1 = pd.DataFrame({'Key' : ['X', 'Y', 'Z'], 'value1' : [1, 2, 3]})
df2 = pd.DataFrame({'Key' : ['X', 'W', 'Z'], 'value2' : [9, 8, 7]})
```

```python
df1
```

```python
df2
```

```python
# merge data frame on key columns

pd.merge(df1, df2, on='Key', how='inner')
```

```python
pd.merge(df1, df2, on='Key', how='outer')
```

```python
pd.merge(df1, df2, on='Key', how='right')
```

```python
pd.merge(df1, df2, on='Key', how='left')
```
