```python
def my_gen():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_gen()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

```python
# reading large files
def read_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'file.txt'

for line in read_file(file_path):
    print(line.strip())
```
