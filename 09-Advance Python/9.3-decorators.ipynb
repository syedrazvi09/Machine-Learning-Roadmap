```python
# function copy
def welcome():
    return 'Welcome to the course'

welcome()
```

```python
wel = welcome
print(wel())

del welcome
print(wel())
```

```python
#closures

def main_welcome():
    msg = 'Hello'
    def sub_welcome():
        print('Welcome to this Part')
        print(msg)

        print('Please learn')

    return sub_welcome()
```

```python
main_welcome()
```

```python
# decorator
def main_welcome(func):
    def sub_welcome():
        print('Welcome to this Part')
        func()

        print('Please learn')

    return sub_welcome()
```

```python
def intro():
    print('This is a ML course')

intro()
```

```python
main_welcome(intro)
```

```python
@main_welcome
def intro():
    print('This is a ML course')
```
